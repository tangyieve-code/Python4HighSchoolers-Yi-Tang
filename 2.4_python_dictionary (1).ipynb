# Python Dictionaries:
A **dictionary** is a Python data structure that stores data as **key–value pairs**.

You can think of it like a real dictionary:

- A **word** is the key.
- Its **definition** is the value.

Dictionaries are useful when each value has a meaningful label, such as a person's name, age, email address, or account status.

````python
student = {
    "name": "John Smith",  #separated by comma ,
    "age": 24,
    "major": "Computer Science"
}

print(student)
print(type(student))
````

## 1. Dictionary syntax

A dictionary is written with curly braces `{}`.

Each entry has the form:

```python
key: value
```

Entries are separated by commas.

Dictionary keys:

- Must be unique.
- Must be immutable, such as strings, numbers, or tuples.
- Are case-sensitive.

Dictionary values can be almost any Python object.

````python
profile = {
    "username": "john123",
    "active": True,
    "score": 92.5,
    "skills": "Python",
    "city": "Seattle",
    "state": "Washington"
}

print(profile)
````

````python
# print in a pretty way
from pprint import pprint

pprint(profile)
````

## 2. Accessing dictionary values

Use square brackets with a key to retrieve a value.

```python
dictionary[key]
```

This raises a `KeyError` if the key does not exist.

````python
student = {
    "name": "John Smith",
    "age": 24,
    "major": "Computer Science"
}

print(student["name"])
print(student["age"])

print()

print("student name is: ", student["name"])
print("student's age is: ", student["age"])
````

Method summary

| Method | Purpose |
|---|---|
| `get(key, default)` | Safely retrieve a value |
| `keys()` | Return all keys |
| `values()` | Return all values |
| `items()` | Return key–value pairs |
| `update(other)` | Add or update multiple entries |
| `setdefault(key, default)` | Get a value or insert a default |
| `pop(key, default)` | Remove a key and return its value |
| `popitem()` | Remove and return the newest pair |
| `copy()` | Create a shallow copy |
| `clear()` | Remove all entries |
| `fromkeys(keys, value)` | Create a dictionary from keys |

### Using `get()`

The `get()` method safely retrieves a value.

```python
dictionary.get(key)
dictionary.get(key, default_value)
```

Unlike square brackets, `get()` does not raise a `KeyError` when the key is missing.

````python
student = {
    "name": "John Smith",
    "age": 24,
    "major": "Computer Science"
}

print(student.get("major"))
print(student.get("email"))
print(student.get("email", "Email not provided")) #safe way to call, in case key not there
````

## Adding and updating entries

Assign a value to a new key to add an entry.

Assign a value to an existing key to update it.

````python
student["email"] = "john@example.com"  # Add a new entry
student["age"] = 25                    # Update an existing entry

print(student)
````

## Checking whether a key exists

Use the `in` and `not in` operators.

These operators check **keys**, not values.

````python
student = {
    "name": "John Smith",
    "age": 24,
    "major": "Computer Science"
}

print("name" in student)
print("phone" in student)
print("phone" not in student)
````

## Main dictionary methods

The following sections demonstrate the methods used most often.

### `keys()`

Returns a view containing all dictionary keys.

````python
print(student.keys())

for key in student.keys():
    print(key)
````

### `values()`

Returns a view containing all dictionary values.

````python
print(student.values())

for value in student.values():
    print(value)
````

### `items()`

Returns each key and value as a tuple.

This is especially useful when looping through a dictionary.

````python
print(student.items())

for key, value in student.items():
    print(f"{key}: {value}")
````

### `update()`

Adds entries from another dictionary or updates existing entries.

````python
student.update({
    "age": 26,
    "graduation_year": 2027,
    "email": "john@gmail.com"
})

print(student)
````

````python
# can also update using assignment
student["age"] = 30
student["graduation_year"] = 2029

print(student)
````

### `pop()`

Removes a specific key and returns its value.

You can provide a default value to avoid a `KeyError` when the key is missing.

````python
student_copy = student.copy()

removed_email = student_copy.pop("email")
missing_phone = student_copy.pop("phone", "No phone number")

print("Removed:", removed_email)
print("Missing key result:", missing_phone)
print(student_copy)
````

### `popitem()`

Removes and returns the most recently added key–value pair.

It raises a `KeyError` if the dictionary is empty.

````python
temporary = {
    "first": 1,
    "second": 2,
    "third": 3
}

removed_pair = temporary.popitem()

print("Removed pair:", removed_pair)
print("Remaining dictionary:", temporary)
````

### `clear()`

Removes every entry from a dictionary.

````python
data = {"a": 1, "b": 2, "c": 3}
data.clear()

print(data)
````

## Removing entries with `del`

`del` removes an entry by key.

Unlike `pop()`, it does not return the removed value.

````python
course = {
    "title": "Python Basics",
    "duration": "6 weeks",
    "level": "Beginner"
}

del course["duration"]

print(course)
````

## Looping through dictionaries

Looping directly over a dictionary gives its keys.

````python
product = {
    "name": "Keyboard",
    "price": 49.99,
    "in_stock": True
}

for key in product:
    print(key)
````

Use `items()` when you need both keys and values.

````python
for key, value in product.items():
    print(f"{key} -> {value}")
````

## Nested dictionaries

A dictionary can contain another dictionary as a value.

````python
employee = {
    "name": "Jordan",
    "department": "Engineering",
    "contact": {
        "email": "jordan@example.com",
        "phone": "555-0100"
    }
}

print(employee["contact"]["email"])
````

You can update a nested value by following each key level.

````python
employee["contact"]["phone"] = "555-0199"
print(employee)
````

## Dictionary comprehensions

A dictionary comprehension is a compact way to create a dictionary.

General form:

```python
{key_expression: value_expression for item in iterable}
```

````python
squares = {number: number ** 2 for number in range(1, 6)}
print(squares)
````

You can also include a condition.

## 10. Important dictionary behaviors

### Keys are unique

If the same key appears more than once, the last value replaces the earlier value.

````python
person = {
    "name": "John",
    "name": "Sam" #the most recent one takes effect
}

print(person)
````

### Keys must be immutable

Strings, numbers, and tuples can be keys.

Lists and dictionaries cannot be keys because they are mutable.

````python
valid_keys = {
    "name": "Alex",
    100: "One hundred",
    (10, 20): "A coordinate"
}

print(valid_keys)
````

The following code would raise a `TypeError`, so it is commented out:

```python
invalid = {
    [1, 2]: "A list cannot be a key"
}
```

### Dictionaries preserve insertion order

When you iterate through a modern Python dictionary, entries appear in the order in which they were added.

````python
ordered_example = {}
ordered_example["first"] = 1
ordered_example["second"] = 2
ordered_example["third"] = 3

print(ordered_example)
````

## Common mistakes

### Mistake 1: Accessing a missing key with square brackets

````python
account = {"username": "alex123"}

# This would raise a KeyError:
print(account["email"])
````

````python
# Safer:
print(account.get("email", "Email not found"))
````

### Mistake 2: Checking for a value with `in`

By default, `in` checks keys.

````python
colors = {
    "warning": "yellow",
    "error": "red"
}

print("red" in colors)            # Checks keys
print("red" in colors.values())   # but when checking values, needs to write .values()
````

## 13. Practical example: Counting words

A dictionary can be used to count how many times each word appears.

````python
sentence = "python is simple and python is powerful"
word_counts = {}

for word in sentence.split():
    word_counts[word] = word_counts.get(word, 0) + 1

print(word_counts)
````
