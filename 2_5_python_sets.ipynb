# Python Sets:

A **set** is a Python data structure that stores a collection of **unique values**.

Sets are useful when you want to:

- Remove duplicate values
- Check whether an item exists
- Compare groups of data
- Find common or different values
- Perform mathematical set operations

This notebook introduces Python sets and their most important methods with runnable examples.

## Creating a set with content

A set is usually created with curly braces `{}`.

Each value appears only once, even if it is written multiple times.

````python
colors = {"red", "blue", "green"}

print(colors)
print(type(colors))
````

````python
numbers = {1, 2, 2, 3, 3, 3}

print(numbers)  # Duplicate values are removed
````

## Creating an empty set

Use `set()` to create an *empty* set.

Do not use `{}` because `{}` creates an empty dictionary.

````python
empty_set = set()
empty_dictionary = {}

print(type(empty_set))
print(type(empty_dictionary))

print()

print(empty_set)
print(empty_dictionary)
````

## Access a set

A Python set:

- Stores only unique values
- Is unordered
- Is mutable
- Does not support indexing
- Can contain immutable values such as strings, numbers, and tuples

Because sets are unordered, their displayed order may vary.

````python
languages = {"Python", "Java", "JavaScript"}

print(languages)
````

````python
# This would cause a TypeError because sets do not support indexing:
# print(languages[0])
````

If set doesn't support indexing (becuase it is unordered). How to access it?

````python
# my_set = {"apple", "banana", "cherry"}
# my_set[0]
````

````python
# through iterating over the set
my_set = {"apple", "banana", "cherry"}

for item in my_set:
    print(item)
````

````python
# convert set to a list, then using index to access it
# but since set is unordered, so the order may be different than the order you created it
my_set = {"apple", "banana", "cherry"}

my_list = list(my_set)

print(my_list[0])
print(my_list[1])
````

## 4. Creating a set from another collection

The `set()` constructor can convert lists, tuples, strings, and other iterable objects into sets.

This is commonly used to remove duplicates from a list.

````python
scores = [90, 80, 90, 70, 80, 100]

unique_scores = set(scores)

print("Original list:", scores)
print("Unique scores:", unique_scores)
````

````python
letters = set("banana")

print(letters)
````

## Checking membership

Use `in` to check whether a value exists in a set.

Use `not in` to check whether it does not exist.

````python
languages = {"Python", "Java", "JavaScript"}

print("Python" in languages)
print("C++" in languages)
print("C++" not in languages)
````

## Method summary

|---|---|

| `intersection(other)` | Return shared values |

### `add()`

The `add()` method adds one value to a set.

If the value already exists, the set does not change.

````python
fruits = {"apple", "banana"}

fruits.add("orange")
fruits.add("apple")  # Already exists

print(fruits)
````

### `update()`

The `update()` method adds multiple values from another iterable, such as a list, tuple, or set.

````python
fruits = {"apple", "banana"}

fruits.update(["orange", "mango", "grape"])

print(fruits)
````

````python
fruits.update({"kiwi", "pear"})

print(fruits)
````

### `remove()`

The `remove()` method removes a value.

It raises a `KeyError` if the value does not exist.

````python
fruits = {"apple", "banana", "orange"}

fruits.remove("banana")

print(fruits)

# This would raise a KeyError:
# fruits.remove("watermelon")
````

### `discard()`

The `discard()` method also removes a value.

Unlike `remove()`, it does not raise an error when the value is missing.

````python
fruits = {"apple", "banana", "orange"}

fruits.discard("banana")
fruits.discard("watermelon")  # No error

print(fruits)
````

### `pop()`

The `pop()` method removes and returns an arbitrary value.

Because sets are unordered, you should not depend on which value is removed.

````python
fruits = {"apple", "banana", "orange"}

removed_fruit = fruits.pop()

print("Removed:", removed_fruit)
print("Remaining:", fruits)
````

### `clear()`

The `clear()` method removes all values from a set.

````python
fruits = {"apple", "banana", "orange"}

fruits.clear()

print(fruits)
````

### `copy()`

The `copy()` method creates a shallow copy of a set.

````python
original = {"Python", "Java", "C++"}
copied = original.copy()

copied.add("JavaScript")

print("Original:", original)
print("Copy:", copied)
````

## Set operations

Assume that we have two groups of programming skills.

````python
# backend_skills = {"Python", "Java", "SQL"}
# frontend_skills = {"JavaScript", "HTML", "CSS", "Python"}
backend_skills = {1, 2, 3, 4}
frontend_skills = {3, 4, 5, 6}

print("Backend:", backend_skills)
print("Frontend:", frontend_skills)
````

### `union()`

The `union()` method returns all unique values from both sets.

The `|` operator performs the same operation.

image.png

````python
all_skills_method = backend_skills.union(frontend_skills)
all_skills_operator = backend_skills | frontend_skills

print(all_skills_method)
print(all_skills_operator)
````

### `intersection()`

The `intersection()` method returns values that appear in both sets.

The `&` operator performs the same operation.

````python
shared_skills_method = backend_skills.intersection(frontend_skills)
shared_skills_operator = backend_skills & frontend_skills

print(shared_skills_method)
print(shared_skills_operator)
````

### `difference()`

The `difference()` method returns values that appear in the first set but not the second.

The `-` operator performs the same operation.

The order matters.

````python
backend_only = backend_skills.difference(frontend_skills)
frontend_only = frontend_skills - backend_skills

print("Backend only:", backend_only)
print("Frontend only:", frontend_only)
````

### `symmetric_difference()`

The `symmetric_difference()` method returns values that appear in either set, but not in both.

The `^` operator performs the same operation.

````python
different_skills_method = backend_skills.symmetric_difference(frontend_skills)
different_skills_operator = backend_skills ^ frontend_skills

print(different_skills_method)
print(different_skills_operator)
````

## Methods that modify the original set

The following methods update a set directly:

- `intersection_update()`
- `difference_update()`
- `symmetric_difference_update()`

### `intersection_update()`

Keeps only values that also appear in another set.

````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

set_a.intersection_update(set_b)

print(set_a)
````

### `difference_update()`

Removes values that also appear in another set.

````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

set_a.difference_update(set_b)

print(set_a)
````

### `symmetric_difference_update()`

Keeps values that appear in one set but not both.

````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

set_a.symmetric_difference_update(set_b)

print(set_a)
````

## Comparing sets

### `issubset()`

Returns `True` if every value in one set also exists in another set.

The `<=` operator can also be used.

````python
required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "Java"}

print(required_skills.issubset(candidate_skills))
print(required_skills <= candidate_skills)
````

### `issuperset()`

Returns `True` if one set contains every value in another set.

The `>=` operator can also be used.

````python
print(candidate_skills.issuperset(required_skills))
print(candidate_skills >= required_skills)
````

### `isdisjoint()`

Returns `True` when two sets have no values in common.

````python
group_a = {"Alice", "Bob"}
group_b = {"Sam", "Jordan"}
group_c = {"Bob", "Taylor"}

print(group_a.isdisjoint(group_b))
print(group_a.isdisjoint(group_c))
````

## Set comprehensions

A set comprehension is a compact way to create a set.

General form:

```python
{expression for item in iterable}
```

````python
squares = {number ** 2 for number in range(1, 6)}

print(squares)
````

You can also add a condition.

````python
even_squares = {
    number ** 2
    for number in range(1, 11)
    if number % 2 == 0
}

print(even_squares)
````

## Immutable sets with `frozenset`

A regular set is mutable, which means it can change.

A `frozenset` is immutable, so values cannot be added or removed after creation.

````python
permissions = frozenset({"read", "write"})

print(permissions)
print(type(permissions))

# This would raise an AttributeError:
# permissions.add("delete")
````

## Practical example: Removing duplicates

A common use of sets is removing duplicate values from a list.

````python
names = ["Alex", "Sam", "Alex", "Jordan", "Sam", "Taylor"]

unique_names = set(names)

print("Original:", names)
print("Unique:", unique_names)
````

If you need the final result as a list, convert it back with `list()`.

Remember that the original order is not guaranteed.

````python
unique_names_list = list(set(names))

print(unique_names_list)
````

## Practical example: Comparing course enrollment

Sets make it easy to compare students enrolled in different courses.

````python
python_students = {"Alex", "Jordan", "Sam", "Taylor"}
java_students = {"Sam", "Taylor", "Morgan"}

both_courses = python_students & java_students
python_only = python_students - java_students
all_students = python_students | java_students

print("Both courses:", both_courses)
print("Python only:", python_only)
print("All students:", all_students)
````

## 15. Common mistakes

### Mistake 1: Using `{}` for an empty set

```python
empty = {}       # Dictionary
empty = set()    # Set
```

### Mistake 2: Trying to use indexing

Sets do not support expressions such as `my_set[0]`.

### Mistake 3: Adding a mutable value

Lists and dictionaries cannot be set elements because they are mutable.

### Mistake 4: Depending on element order

A set is unordered, so do not use it when element order is important.

````python
valid_set = {
    "Python",
    42,
    (10, 20)
}

print(valid_set)

# This would raise a TypeError:
# invalid_set = {[1, 2, 3]}
````
