## 1. What is a tuple?

A **tuple** is an ordered collection of values. Tuples are usually written inside parentheses.

A tuple is an immutable, ordered collection of elements that can store heterogeneous data types such as strings, integers, and more.

A tuple is a linear data structure where elements are stored sequentially, similar to arrays or lists, but unlike lists, tuples cannot be modified after creation.

````python
student = ("Maya", 16, 91.5, True)

print(student)
print(type(student))
````

The tuple above contains:

- a string: the student's name;
- an integer: the student's age;
- a float: the student's average score;
- a Boolean: whether the student is currently enrolled.


## 2. Creating tuples

Tuples can contain one value, several values, or no values.


````python
empty_tuple = ()
single_item_tuple = ("Python",)
multiple_item_tuple = ("Math", "Science", "History")

print(empty_tuple)
print(single_item_tuple)
print(multiple_item_tuple)
print(type(single_item_tuple))
````

### The comma matters

A one-item tuple must include a comma. Without the comma, Python treats the value as an ordinary string, integer, or other object.


````python
not_a_tuple = ("Python")
is_a_tuple = ("Python",)

print(type(not_a_tuple))
print(type(is_a_tuple))
````

Parentheses are often used for readability, but commas are what actually create a tuple.


````python
coordinates = 34.05, -118.24

print(coordinates)
print(type(coordinates))
````

## 3. Tuple indexing

Tuples are ordered, so each value has an index. Python starts counting at index `0`.

Negative indexes count backward from the end.


````python
subjects = ("Math", "Science", "History", "Art")

print(subjects[0])
print(subjects[1])
print(subjects[-1])
print(subjects[-2])
````

## 4. Tuple slicing

Slicing retrieves a section of a tuple. The ending index is not included.


````python
subjects = ("Math", "Science", "History", "Art", "Music")

print(subjects[1:4])
print(subjects[:3])
print(subjects[2:])
print(subjects[::-1]) #Python slicing, sequence[start:stop:step], if step is -1, start is from the end
````

## 5. Tuples are immutable

Tuples are **immutable**, which means their elements cannot be changed, added, or removed after the tuple is created.


````python
coordinates = (34.05, -118.24)

print(coordinates)
````

The following code would cause a `TypeError` because tuple items cannot be reassigned:

```python
coordinates[0] = 35.00
```

Lists are mutable, but tuples are immutable.


### When are tuples useful?

Tuples are useful when values belong together and should not be changed accidentally.

Examples include:

- geographic coordinates;
- RGB color values;
- dates;
- dimensions;
- fixed settings;
- records returned from a function.


````python
location = (37.7749, -122.4194)
screen_resolution = (1920, 1080)
rgb_color = (255, 165, 0)
event_date = (2026, 10, 15)

print("Location:", location)
print("Screen resolution:", screen_resolution)
print("RGB color:", rgb_color)
print("Event date:", event_date)
````

## 6. Common tuple operations

Tuples support several operations that are also available for strings and lists.

| Operation | Example | Purpose |
|---|---|---|
| Length | `len(values)` | Counts tuple elements |
| Membership | `value in values` | Checks whether a value exists |
| Concatenation | `tuple1 + tuple2` | Combines tuples |
| Repetition | `values * 3` | Repeats tuple contents |
| Indexing | `values[0]` | Retrieves one element |
| Slicing | `values[1:3]` | Retrieves part of a tuple |
| Minimum | `min(values)` | Finds the smallest value |
| Maximum | `max(values)` | Finds the largest value |
| Sum | `sum(values)` | Adds numeric values |


````python
scores = (88, 92, 79, 95)

print("Length:", len(scores))
print("Contains 92:", 92 in scores)
print("Lowest score:", min(scores))
print("Highest score:", max(scores))
print("Total score:", sum(scores))
print("Average score:", sum(scores) / len(scores))
````

## 7. Built-in tuple methods

Tuples have only two built-in methods:

- `count()`
- `index()`

Tuples have fewer methods than lists because tuples cannot be changed.


### The `count()` method

The `count()` method returns the number of times a value appears in a tuple.


````python
grades = ("A", "B", "A", "C", "A", "B")

number_of_a_grades = grades.count("A")
number_of_b_grades = grades.count("B")

print("Number of A grades:", number_of_a_grades)
print("Number of B grades:", number_of_b_grades)
````

### The `index()` method

The `index()` method returns the index of the first matching value.


````python
courses = ("Math", "Science", "History", "Science")

science_position = courses.index("Science")
history_position = courses.index("History")

print("First Science index:", science_position)
print("History index:", history_position)
````

If the requested value is not in the tuple, `index()` raises a `ValueError`. It is often helpful to check membership first.


````python
courses = ("Math", "Science", "History")
search_value = "Art"

if search_value in courses:
    print(courses.index(search_value))
else:
    print(f"{search_value} is not in the tuple.")
````

## 8. Tuple unpacking

Tuple unpacking assigns tuple values to separate variables.


````python
student_record = ("Maya Johnson", 16, 91.5)

student_name, student_age, average_score = student_record

print("Name:", student_name)
print("Age:", student_age)
print("Average:", average_score)
````

The number of variables must normally match the number of tuple elements.


### Extended unpacking

The `*` operator can collect several remaining values into a list.


````python
scores = (88, 92, 79, 95, 84)

first_score, *middle_scores, last_score = scores

print("First score:", first_score)
print("Middle scores:", middle_scores)
print("Last score:", last_score)
````

## 9. Returning tuples from functions

A function can return several values. Python groups those values into a tuple.


````python
def calculate_score_summary(scores):
    """Return the lowest, highest, and average score."""
    lowest = min(scores)
    highest = max(scores)
    average = sum(scores) / len(scores)
    return lowest, highest, average

summary = calculate_score_summary((88, 92, 79, 95))

print(summary)
print(type(summary))
````

The returned tuple can be unpacked into separate variables.


````python
lowest_score, highest_score, average_score = calculate_score_summary((88, 92, 79, 95))

print(f"Lowest: {lowest_score}")
print(f"Highest: {highest_score}")
print(f"Average: {average_score:.2f}")
````

## 10. Tuples and other data structures

Tuples can be stored inside lists and dictionaries. Lists can also be stored inside tuples.


````python
students = [
    ("Maya", 91.5),
    ("Noah", 87.0),
    ("Liam", 93.5),
]

for name, average in students:
    print(f"{name}: {average}")
````

````python
school_locations = {
    "Central High": (34.05, -118.24),
    "North High": (34.18, -118.31),
}

print(school_locations["Central High"])
````

## 11. Converting between lists and tuples

The `tuple()` function converts another iterable into a tuple. The `list()` function converts a tuple into a list.


````python
course_list = ["Math", "Science", "History"]
course_tuple = tuple(course_list)

print(course_tuple)
print(type(course_tuple))

editable_courses = list(course_tuple)
editable_courses.append("Art")

print(editable_courses)
````

This conversion is useful when data needs to be edited temporarily and then stored as a fixed tuple.


````python
coordinates = (34.05, -118.24)

editable_coordinates = list(coordinates)
editable_coordinates[0] = 34.10
updated_coordinates = tuple(editable_coordinates)

print("Original:", coordinates)
print("Updated:", updated_coordinates)
````
